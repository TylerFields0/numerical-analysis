```python
# EX 0.1.6
# a. p(x) = a0 + a5x^5 + a10x^10 + a15x^15

# Reference
import numpy as np
a0 = -13; a5 = 26; a10 = 94; a15 = -2
x = np.array([-2, -1, 0, 1, 2])
px0 = a0 + a5 * (x**5) + a10 * (x**10) + a15 * (x**15)
print(px0)


# Implementation 1
x2 = x * x
x3 = x2 * x
x5 = x3 * x2
x10 = x5 * x5
x15 = x10 * x5

px1 = a0 + a5 * x5 + a10 * x10 + a15 * x15
print(px1)

# 8 multiplications, 3 additions, 0 power operators

# Implementation 2 (nested)
px3 = a0 + x5 * (a5 + x5 * (a10 + a15 * x5))
print(px3)

# 6 multiplications, 3 additions, 0 power operators
```

```python
# EX 0.1.6
# b. p(x) = a7x^7 + a12x^12 + a17x^17 + a22x^22 + a27x^27

# Reference
import numpy as np
a7 = 5; a12 = 2; a17 = 4; a22 = -2; a27 = 6
x = np.array([-2, -1, 0, 1, 2])
px0 = a7 * x**7 + a12 * x**12 + a22 * x**22 + a27 * x**27
print(px0)

#Implementation 1
x2 = x * x
x3 = x2 * x
x5 = x3 * x2
x7 = x5 * x2
x10 = x5 * x5

px1 = x7 * (a7 + x5 * (a12 + x10 * (a22 + a27 * x5)))
print(px1)

# 9 multiplications, 3 addititions, 0 power operators
```

```python
# CP.0.1.2
# Use nest to evaluate p(x) = 1 - x + x^2 - x^3 + ... + x^98 - x^99 at x = 1.00001

import numpy as np

def polyval_nested(c, x):
  d = np.size(c)
  px = c[d - 1]
  for i in range(d - 2, -1, -1 ):
    px = px * x + c[i]
  return px

x = 1.00001
d = 99
c = np.ones([d+1])
px1 = polyval_nested(c, x)
print(px1)

# Simplified version
px2 = ((x ** (d + 1)) - 1) / (x - 1)
print(px2)
```
