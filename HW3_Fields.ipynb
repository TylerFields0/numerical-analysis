```python
import numpy as np
import copy
from math import sqrt
import scipy
from scipy.optimize import fsolve
from scipy.optimize import root
```

```python
# ----------------------
# Exercise CP.2.2.1.b2c
# ----------------------

#
# useage:
#           L, U = lu_no_pivoting( A )
#
# performs LU_no_pivoting on `A`
#
# input:  A:       n-by-n matrix to be factored
# output: L an U:  the L and U factors defined by:
#                  (1) L is n-by-n lower unit triangular
#                  (2) U is n-by-n upper triangular
#                  (3) A = L * U
#
# warning: no pivoting is used, this algorithm will exit with error message if
# an exact-zero pivot is encountered, this code might be quite unstable when
# small pivots are encountered during the factorization.
#
# this is the standard in-place algorithm (so store L and U on top of A), but
# there is a deep copy of A so that the ``input A`` is not overwritten, in addition,
# we rewrite L and U in separate matrices
#
def lu_no_pivoting( A_ ):
  A = copy.deepcopy(A_)
#
  n = A.shape[0]
  for k in range(0,n-1):
    if ( A[k,k] == 0. ): print("oops, zero pivot encountered\n"); break
    for i in range(k+1,n):
      A[i,k] = A[i,k] / A[k,k]
      for j in range(k+1,n):
        A[i,j] = A[i,j] - A[k,j] * A[i,k]
#
  L = np.tril(A,-1)+np.eye(n)
  U = np.triu(A)
  return L, U
```

```python
# (b)
A = np.array([  [  4.,  2.,  0. ],
                [  4.,  4.,  2. ],
                [  2.,  2.,  3. ] ] )

# using our bucket algorithm `lu_no_pivoting`

L, U = lu_no_pivoting( A )

print("L=\n",L)
print("U=\n",U)

# check that A = LU
print("\nL @ U=\n", L @ U)
print("A=\n",A)
print("\n|| A - LU ||_oo / || A ||_oo = ", np.linalg.norm(A - L@U,np.inf)/np.linalg.norm(A,np.inf) )
```

```python
# ---------------------
# Exercise CP.2.2.2.b
# ---------------------

#
# useage:
#           x = backward_substitution( A, b )
#
# performs backward substitution Ax = b where A is upper triangular
#
# input:  A and b: represent the upper triangular linear system of equations
#                  Ax=b we want to solve
# output: x:       solution of x of Ax=b (this is done in place)
#
# this is the standard in-place algorithm (so store x on top of b), but
# there is a deep copy of b so that the ``input b`` is not overwritten
#
def backward_substitution( A, b_ ):
  b = copy.deepcopy(b_)
  n = A.shape[0]
  for i in range(n-1,-1,-1):
    for j in range(i+1,n):
      b[i] = b[i] - A[i,j] * b[j]
    b[i] = b[i] / A[i,i]
  return b

# useage:
#           x = forward__substitution( A, b )
#
# performs forward substitution Ax = b where A is lower triangular
#
# input:  A and b: represent the lower triangular linear system of equations
#                 Ax=b we want to solve
# output: x:       solution of x of Ax=b (this is done in place)
#
# this is the standard in-place algorithm (so store x on top of b), but
# there is a deep copy of b so that the ``input b`` is not overwritten
#
def forward__substitution( A, b_ ):
  b = copy.deepcopy(b_)
  n = A.shape[0]
  for i in range(0,n,):
    for j in range(0,i):
      b[i] = b[i] - A[i,j] * b[j]
    b[i] = b[i] / A[i,i]
  return b

A = np.array([  [  4.,  2.,  0. ],
                [  4.,  4.,  2. ],
                [  2.,  2.,  3. ] ] )

b = np.array([  [  2. ],
                [  4. ],
                [  6. ] ] )

# solve Ax=b with np.linalg.solve
# and check that Ax is b

x = np.linalg.solve(A, b)

print("With np.linalg.solve, we find x = \n", x)

print("\nWe check that Ax is b. Indeed [ Ax, b ] =\n", np.concatenate((A@x, b), axis=1) )

print("\nrelative backward error: || b - Ax  ||_oo / || b ||_oo =\n", np.linalg.norm(b-A@x,np.inf)/np.linalg.norm(b,np.inf) )

# using our bucket algorithms
#   `lu_no_pivoting`
#   `forward__substitution`
#   `backward_substitution`

L, U = lu_no_pivoting( A )
y = forward__substitution( L, b )
x = backward_substitution( U, y )

print("With our algorithms, we find\n")
print("y=\n",y)
print("x=\n",x)

print("\nWe check that Ax is b. Indeed [ Ax, b ] =\n", np.concatenate((A@x, b), axis=1) )

print("\nrelative backward error: || b - Ax  ||_oo / || b ||_oo =\n", np.linalg.norm(b-A@x,np.inf)/np.linalg.norm(b,np.inf) )
```

```python
def lu_no_pivoting( A_, inplace = False ):
#
  if ( inplace ):
    A = A_
  else:
    A = copy.deepcopy(A_)
# begin main code
  n = A.shape[0]
  for k in range(0,n-1):
    if ( A[k,k] == 0. ): print("oops, zero pivot encountered\n"); break
    for i in range(k+1,n):
      A[i,k] = A[i,k] / A[k,k]
      for j in range(k+1,n):
        A[i,j] = A[i,j] - A[k,j] * A[i,k]
# end main code
  if not inplace:
    L = np.tril(A,-1)+np.eye(n)
    U = np.triu(A)
    return L, U

def backward_substitution( A, b_, inplace = False ):
#
  if ( inplace ):
    b = b_
  else:
    b = copy.deepcopy(b_)
#
  n = A.shape[0]
  for i in range(n-1,-1,-1):
    for j in range(i+1,n):
      b[i] = b[i] - A[i,j] * b[j]
    b[i] = b[i] / A[i,i]
#
  if not inplace:
    return b

def forward__substitution( A, b_, inplace = False, unit = False ):
  if ( inplace ):
    b = b_
  else:
    b = copy.deepcopy(b_)
#
  n = A.shape[0]
  for i in range(0,n,):
    for j in range(0,i):
      b[i] = b[i] - A[i,j] * b[j]
    if not unit:
      b[i] = b[i] / A[i,i]
#
  if not inplace:
    return b
```
