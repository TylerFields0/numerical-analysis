```python
# -----------------
# Exercise CP.4.1.5
# -----------------

# A company test-markets a new soft drink in 22 cities of approximately equal
# size. The selling price (in dollars) and the number sold per week in the
# cities are listed as follows:
#    -----------------------------
#    | city | price | sales/week |
#    |------|------ |------------|
#    |   1  |  0.59 |    3980    |
#    |   2  |  0.80 |    2200    |
#    |   3  |  0.95 |    1850    |
#    |   4  |  0.45 |    6100    |
#    |   5  |  0.79 |    2100    |
#    |   6  |  0.99 |    1700    |
#    |   7  |  0.90 |    2000    |
#    |   8  |  0.65 |    4200    |
#    |   9  |  0.79 |    2440    |
#    |  10  |  0.69 |    3300    |
#    |  11  |  0.79 |    2300    |
#    |  12  |  0.49 |    6000    |
#    |  13  |  1.09 |    1190    |
#    |  14  |  0.95 |    1960    |
#    |  15  |  0.79 |    2760    |
#    |  16  |  0.65 |    4330    |
#    |  17  |  0.45 |    6960    |
#    |  18  |  0.60 |    4160    |
#    |  19  |  0.89 |    1990    |
#    |  20  |  0.79 |    2860    |
#    |  21  |  0.99 |    1920    |
#    |  22  |  0.85 |    2160    |
#    -----------------------------

# (a)

data = np.array([
   [  1., 0.59, 3980. ],
   [  2., 0.80, 2200. ],
   [  3., 0.95, 1850. ],
   [  4., 0.45, 6100. ],
   [  5., 0.79, 2100. ],
   [  6., 0.99, 1700. ],
   [  7., 0.90, 2000. ],
   [  8., 0.65, 4200. ],
   [  9., 0.79, 2440. ],
   [ 10., 0.69, 3300. ],
   [ 11., 0.79, 2300. ],
   [ 12., 0.49, 6000. ],
   [ 13., 1.09, 1190. ],
   [ 14., 0.95, 1960. ],
   [ 15., 0.79, 2760. ],
   [ 16., 0.65, 4330. ],
   [ 17., 0.45, 6960. ],
   [ 18., 0.60, 4160. ],
   [ 19., 0.89, 1990. ],
   [ 20., 0.79, 2860. ],
   [ 21., 0.99, 1920. ],
   [ 22., 0.85, 2160. ]])

xx = data[:,1]
yy = data[:,2]

A = np.array([ np.ones( xx.shape ), xx ]).T
b = np.array([ yy ]).T
print("A =\n",A)
print("b =\n",b)

x = np.linalg.lstsq(A,b,rcond=None)[0]

print("x =\n",x)
print("[ Ax, b ]=\n",np.concatenate((A@x,b),axis=1))
r = b - A@x
print("r =\n",r)
print("|| b - Ax ||_2 = ", f"{np.linalg.norm( r, 2 ):.4f}")
print("RMSE = ", f"{np.linalg.norm( r, 2 ) / sqrt( A.shape[0] ) :.4f}")
print("|| A^T ( b - Ax ) ||_oo =",
      f"{np.linalg.norm( A.T@( A@x - b), np.inf ):.1e}" )

xxx = np.linspace( 0.3, 1.2, 10)
yyy = x[0] + x[1] * xxx
plt.plot(xxx, yyy, '--b',label=f'{x[1,0]:7.5f}'+' x + '+f'{x[0,0]:7.5f}')
plt.plot( xx, yy, 'ro')
[ plt.text( xx[i], yy[i], f"{data[i,0]:.0f}") for i in range(0,data.shape[0]) ]
plt.legend()
plt.xlabel('price (in dollars, per unit)')
plt.ylabel('sales/week (in number of unit sold)')
plt.xlim([ 0.4, 1.2 ])
plt.ylim([ 0., 7500.])
plt.grid()
plt.show()
```

```python
# (b)

manufacturing_cost = 0.23

x_opt = ( manufacturing_cost - x[0] / x[1] ) / 2.
y_opt = ( x[0] + x[1] * x_opt ) * (x_opt - manufacturing_cost)

print("the optimal selling price is $" + f"{x_opt[0]:6.4f}")
print("the total_profit is $" + f"{y_opt[0]:6.2f}" + " per week")

xxx = np.linspace( 0., 1.4, 100)
total_profit = ( x[0] + x[1] * xxx ) * (xxx - manufacturing_cost)
plt.plot(xxx, total_profit, '--b')
plt.xlim([ manufacturing_cost, 1.4 ])
plt.ylim([ 0., 2400.])
plt.plot( x_opt, y_opt, 'or')
plt.xlabel('price (in dollars, per unit)')
plt.ylabel('total profit (in dollars per week)')
plt.grid()
plt.show()
```

```python
# -------------------
# Exercise CP.4.5.1.b
# -------------------

#	Apply the Gauss--Newton Method to find the point (x,y) for which
# the sum of the squared distance to the three circles below is minimized. Use
# initial vector (x_0, y_0) = (0,0).
#
# Circles with centers (-1,0), (1,1), (1,-1) and all radii 1.

x0 = np.array( [ -1.,  0. ]); R0 = 1.
x1 = np.array( [  1.,  1. ]); R1 = 1.
x2 = np.array( [  1., -1. ]); R2 = 1.

f = lambda x: np.array([
    sqrt( ( x[0] - x0[0] )**2 + ( x[1] - x0[1] )**2 ) - R0,
    sqrt( ( x[0] - x1[0] )**2 + ( x[1] - x1[1] )**2 ) - R1,
    sqrt( ( x[0] - x2[0] )**2 + ( x[1] - x2[1] )**2 ) - R2,
])

J = lambda x : np.array([
    [ ( x[0] - x0[0] ) / sqrt( ( x[0] - x0[0] )**2 + ( x[1] - x0[1] )**2 ),
      ( x[1] - x0[1] ) / sqrt( ( x[0] - x0[0] )**2 + ( x[1] - x0[1] )**2 ) ],
    [ ( x[0] - x1[0] ) / sqrt( ( x[0] - x1[0] )**2 + ( x[1] - x1[1] )**2 ),
      ( x[1] - x1[1] ) / sqrt( ( x[0] - x1[0] )**2 + ( x[1] - x1[1] )**2 ) ],
    [ ( x[0] - x2[0] ) / sqrt( ( x[0] - x2[0] )**2 + ( x[1] - x2[1] )**2 ),
      ( x[1] - x2[1] ) / sqrt( ( x[0] - x2[0] )**2 + ( x[1] - x2[1] )**2 ) ],
      ])

x = np.array( [ 0., 0. ])

fx = f( x )

for _ in range(0,10):
  Jx = J(x)
  JxTJx = Jx.T @ Jx
  x = x - np.linalg.solve( JxTJx, Jx.T @ fx )
  fx = f(x)
  rmse = np.linalg.norm( fx, 2 ) / sqrt( 3. )
  print( "x = [ ", f"{x[0]:.10f}", ",", f"{x[1]:.10f}", " ]", "    rmse = ", f"{rmse:.16f}")

ttt = np.linspace( 0., 2.*pi, 100)
xxx = R0 * np.cos( ttt ) + x0[0]
yyy = R0 * np.sin( ttt ) + x0[1]
plt.plot(xxx, yyy, '--b')
plt.plot(  x0[0], x0[1], 'bo')

ttt = np.linspace( 0., 2.*pi, 100)
xxx = R1 * np.cos( ttt ) + x1[0]
yyy = R1 * np.sin( ttt ) + x1[1]
plt.plot(xxx, yyy, '--b')
plt.plot(  x1[0], x1[1], 'bo')

ttt = np.linspace( 0., 2.*pi, 100)
xxx = R2 * np.cos( ttt ) + x2[0]
yyy = R2 * np.sin( ttt ) + x2[1]
plt.plot(xxx, yyy, '--b')
plt.plot(  x2[0], x2[1], 'bo')

plt.plot(  x[0], x[1], 'ro')

plt.xlim([ -1.5, 2.5 ])
plt.ylim([ -2.5, 2.5 ])
plt.grid()
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

```python
# -----------------
# Exercise CP.4.5.8
# -----------------

#	Use the Levenberg–Marquardt Method with lambda = 1 to fit a blood concentration
# model to the data of Example 4.11 without linearization

# The power law model is W = c_0 H ^ c_1.
#
# Example 4.11 data
#
#    |--------------------------------|
#    |  hour  | concentration (ng/ml) |
#    |--------|-----------------------|
#    |   1    |         8.0           |
#    |   2    |        12.3           |
#    |   3    |        15.5           |
#    |   4    |        16.8           |
#    |   5    |        17.1           |
#    |   6    |        15.8           |
#    |   7    |        15.2           |
#    |   8    |        14.0           |
#    |--------|-----------------------|

data = np.array([
  [  1,  8.0 ],
  [  2, 12.3 ],
  [  3, 15.5 ],
  [  4, 16.8 ],
  [  5, 17.1 ],
  [  6, 15.8 ],
  [  7, 15.2 ],
  [  8, 14.0 ]])

hour = data[:,0]
concentration = data[:,1]

f = lambda c: c[0] * hour ** c[1] - concentration

J = lambda c: np.array([
    hour ** c[1],
    c[0] * np.log(hour) * hour ** c[1]
]).T

c = np.array( [ 16., 2. ])
lbda = 1

fc = f( c )

for _ in range(0,200):
  Jc = J(c)
  JcTJc = np.transpose(Jc) @ Jc
  JcTJc_plI = JcTJc + lbda * np.eye(np.shape(Jc)[1])
  JcTJc_plD = JcTJc + lbda * np.diag(np.diag(JcTJc))

  # Gauss Netwon uses JcTJc
  # Levenberg uses JcTJc_plI
  # Levenberg Marquardt uses JcTJc_plD
  c = c - np.linalg.solve( JcTJc_plD, np.transpose(Jc) @ fc )
  fc = f(c)
  nrm_r = np.linalg.norm( fc, 2 )
  print( "x = [ ", f"{c[0]:.10f}", ",", f"{c[1]:.10f}", " ]", "    nrm_r = ", f"{nrm_r:.16f}")

print("RMSE = ", f"{np.linalg.norm( fc, 2 ) / sqrt( hour.shape[0] ) :.4f}")

xxx = np.linspace( 0., 8.5, 100)
yyy = np.zeros(xxx.shape[0])
for i in range(0,xxx.shape[0]):
  yyy[i] = c[0] * xxx[i] ** c[1]

plt.plot(xxx, yyy, '--b',label=f'W = '+f'{c[0]:7.5f}'+' * H ^ '+f'{c[1]:7.5f}' )
plt.plot( hour, concentration, 'ro')
[ plt.text( hour[i], concentration[i], f"{data[i,0]:.0f}") for i in range(0,data.shape[0]) ]
plt.legend()
plt.xlabel('hour')
plt.ylabel('concentration (ng/ml)')
plt.xlim([ 0.,  9 ])
plt.ylim([ 0., 20.])
plt.grid()
plt.show()
```

```python
# --------------------
# Exercise CP.4.5.11.b
# --------------------

#	Apply Levenberg-Marquardt to fit the model y = c1 . e^{− c2 . t} cos( c3 . t + c4 )
# to the following data points, with an appropriate initial guess. State the
# initial guess, the regularization parameter λ used, and the RMSE.
# Plot the best least squares curve and the data points. This problem has
# multiple solutions with the same RMSE, since c4 is only determined modulo 2π.
#
# (ti,yi) ={(1,2), (3,6), (4,4), (5,2), (6,−1), (8,−3)}.}

data = np.array([
   [  1.,  2. ],
   [  3.,  6. ],
   [  4.,  4. ],
   [  5.,  2. ],
   [  6., -1. ],
   [  8., -3. ]])

t = data[:,0]
y = data[:,1]

f = lambda c: c[0] * np.exp( -c[1]*t ) * np.cos( c[2]*t + c[3] ) - y

J = lambda c: np.array([
    np.exp(-c[1]*t) * np.cos(c[2]*t + c[3]),
    -c[0]*t*np.exp(-c[1]*t) * np.cos(c[2]*t + c[3]),
    -c[0]*t*np.exp(-c[1]*t) * np.sin(c[2]*t + c[3]),
    -c[0]*np.exp(-c[1]*t) * np.sin(c[2]*t + c[3])
]).T

c = np.array( [ 5., 0.5, 0.8, -1. ])
lbda = 10.

fc = f( c )

for _ in range(0,300):
  Jc = J(c)
  JcTJc = np.transpose(Jc) @ Jc
  JcTJc_plI = JcTJc + lbda * np.eye(np.shape(Jc)[1])
  JcTJc_plD = JcTJc + lbda * np.diag(np.diag(JcTJc))

  # Gauss Netwon uses JcTJc
  # Levenberg uses JcTJc_plI
  # Levenberg Marquardt uses JcTJc_plD
  c = c - np.linalg.solve( JcTJc_plD, np.transpose(Jc) @ fc )
  fc = f(c)
  nrm_r = np.linalg.norm( fc, 2 )
  print( "x = [ ", f"{c[0]:.10f}", ",", f"{c[1]:.10f}", ",", f"{c[2]:.10f}", ",", f"{c[3]:.10f}", " ]", "    nrm_r = ", f"{nrm_r:.16f}")

print("RMSE = ", f"{np.linalg.norm( fc, 2 ) / sqrt( t.shape[0] ) :.4f}")

xxx = np.linspace( 0., 10., 200)
yyy = np.zeros(xxx.shape[0])
for i in range(0,xxx.shape[0]):
  yyy[i] = c[0] * exp( - c[1] * xxx[i] ) * cos( c[2] * xxx[i] + c[3] )

plt.plot(xxx, yyy, '--b',label=f'{c[0]:7.5f}'+' * e^( - '+f'{c[1]:7.5f}'+' * t ) * cos( '+f'{c[2]:7.5f}'+' * t + '+f'{c[3]:7.5f}'+' )' )
plt.plot(xxx, yyy, '--b' )
plt.plot( t, y, 'ro')
[ plt.text( t[i], y[i], f"{data[i,0]:.0f}") for i in range(0,data.shape[0]) ]
plt.legend()
plt.xlim([ 0., 10. ])
plt.ylim([ -4., 8.])
plt.grid()
plt.show()
```
