```python
import numpy as np
import scipy.integrate
import scipy.optimize
```

```python
x  = lambda t: 0.5 + 0.3*t + 3.9*t**2 - 4.7*t**3
y  = lambda t: 1.5 + 0.3*t + 0.9*t**2 - 2.7*t**3

dfdx = lambda t: 0.3 + 7.8*t - 14.1*t**2
dfdy = lambda t: 0.3 + 1.8*t - 8.1*t**2
```

```python
# Question #1:
# Create a function f(t) = sqrt ( x’(t)^2 + y’(t)^2 ) where x(t) and y(t) are given
# above and x’(t) is the derivative of x(t) with respect to t and y’(t) is the derivative
# of y(t) with respect to t. Then compute L the integral of f(t) from t = 0 to t=1.
# For example, you can use: scipy.integrate.quadrature(). Or you can use any method seen
# in Chapter 5. Give your answer to 6 significant decimal places. Hint: The answer is L=2.49525.

f = lambda t: np.sqrt( dfdx(t)**2 + dfdy(t)**2 )

L = scipy.integrate.quad( f, 0., 1. )[0]
print("L =", f"{L:.6f}")
print()
```

```python
# Question #2:
# Find the value of t_ such that the integral of f(t) from t = 0 to t=t_ is s=0.4
# using a root finding method. For example, you can use: scipy.optimize.root_scalar().
# Or you can use any method seen in Chapter 1. Give your answer to 6 significant
# decimal places. Hint: The answer is t_=0.350239.

s = 0.4

solution = scipy.optimize.root_scalar(
  lambda t_: scipy.integrate.quad( f, 0., t_ )[0] - s,
  bracket=[0., 1.])

t_ = solution.root
print("t_ =", f"{t_:.6f}")
```

```python
# Question #3:
# Repeat Question 2 with the following 24 values of s:
# s = 0.1, 0.2, 0.3, 0.4, . . .  2.2, 2.3 and 2.4.

s_values = np.round(np.arange(0.1, 2.41, 0.1), 1)

t_values = []
for s in s_values:
  solution = scipy.optimize.root_scalar(
  lambda t_: scipy.integrate.quad( f, 0., t_ )[0] - s,
  bracket=[0., 1.])
  t_values.append(solution.root)
  print(f"s = {s:4.1f}    t_ = {solution.root:.6f}")
```

```python
# Question #4:
# Now solve the same problem as in Question 2 but this times using Newton’s method.
# At each iteration of Newton’s method, display the forward error, that is the absolute value of
# the difference between your current iterate and the solution (as computed in Question 2).
# Use the solution of Question 2 with as computed with 16-digit accuracy.
# (Do not use the solution with only 6-digit accuracy.)

def g(t, s):
    return scipy.integrate.quad( f, 0., t )[0] - s

solution = scipy.optimize.root_scalar( g, args=(0.4,), bracket=[0., 1.])
t_true = solution.root
print("t_true =", f"{t_true:.16f}")
print()

t = 0.5   # initial guess
for i in range(0,5):
  t = t - g(t, 0.4) / f(t)
  forward_error = abs( t_true - t )
  print( f"{i+1:2d}", f"{t:.16f}", f"{forward_error:.2e}" )
```
